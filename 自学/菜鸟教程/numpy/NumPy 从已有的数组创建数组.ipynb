```python
import numpy as np

x = [1, 2, 3]
a = np.asarray(x)
print(a)
```

```python
x = (1, 2, 3)
a = np.asarray(x)
print(a)
```

```python
# 方法1：使用dtype=object创建对象数组
a = np.asarray(x, dtype=object)
print(a)

# 方法2：确保所有序列长度一致
x = [(1,2,3), (4,5,6)]  # 改为长度相同
a = np.asarray(x)

# 方法3：用None填充缺失位置
x = [(1,2,3), (4,5,None)]
a = np.array(x, dtype=float)
print(a)
```

```python
import numpy as np 
 
# 使用 range 函数创建列表对象  
list=range(5)
it=iter(list)
 
# 使用迭代器创建 ndarray 
x=np.fromiter(it, dtype=float)
print(x)
```
