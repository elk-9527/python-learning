6、（练习 10.13）有两台车床生产同一种型号的滚珠，根据已往经验可以认为，这两台车床生产的滚珠的直径均服从正态分布。现从这两台车床生产的滚环中分别抽出 8 个和 9 个，测得滚珠的直径（单位为 mm）为

甲车床：15.0，14.5，15.2，15.5，14.8，15.1，15.2，14.8

乙车床：15.2，15.0，14.8，15.2，15.0，15.0，14.8，15.1，14.8

判断乙车床生产的滚珠直径的方差是否与甲车床生产的滚珠直径的方差相等，α=0.05。


```python
import numpy as np
from scipy.stats import f

# 定义数据
x1 = np.array([15.0, 14.5, 15.2, 15.5, 14.8, 15.1, 15.2, 14.8])  # 甲车床数据
x2 = np.array([15.2, 15.0, 14.8, 15.2, 15.0, 15.0, 14.8, 15.1, 14.8])  # 乙车床数据

# 计算样本方差（ddof=1表示除以n-1，得到样本方差）
s1_sq = np.var(x1, ddof=1)  # 甲车床样本方差
s2_sq = np.var(x2, ddof=1)  # 乙车床样本方差

# 确定F统计量（较大方差除以较小方差，确保F≥1）
if s1_sq > s2_sq:
    F_stat = s1_sq / s2_sq
    df1 = len(x1) - 1  # 分子自由度（甲车床样本量-1）
    df2 = len(x2) - 1  # 分母自由度（乙车床样本量-1）
else:
    F_stat = s2_sq / s1_sq
    df1 = len(x2) - 1  # 分子自由度（乙车床样本量-1）
    df2 = len(x1) - 1  # 分母自由度（甲车床样本量-1）

# 显著性水平
alpha = 0.05

# 计算临界值（双侧检验）
F_upper = f.ppf(1 - alpha/2, df1, df2)  # 上侧临界值（α/2分位数）
F_lower = f.ppf(alpha/2, df1, df2)      # 下侧临界值（1-α/2分位数）

# 决策：比较F统计量与临界值
if F_stat > F_upper or F_stat < F_lower:
    print("拒绝原假设：两车床方差不相等")
else:
    print("不拒绝原假设：两车床方差相等")

# 可选：计算p值（双侧检验）
p_value = 2 * min(f.cdf(F_stat, df1, df2), 1 - f.cdf(F_stat, df1, df2))
print(f"F统计量：{F_stat:.4f}")
print(f"上侧临界值：{F_upper:.4f}")
print(f"下侧临界值：{F_lower:.4f}")
print(f"p值：{p_value:.4f}")
```

要解决“判断乙车床生产的滚珠直径方差是否与甲车床相等”的问题，需使用**两个正态总体方差的假设检验（F检验）**，步骤如下：
### **1. 提出假设**
- 原假设 \( H_0: \sigma_1^2 = \sigma_2^2 \)（两车床方差相等）  
- 备择假设 \( H_1: \sigma_1^2 \neq \sigma_2^2 \)（两车床方差不等，双侧检验）  
### **2. 计算样本方差**
#### （1）甲车床（样本1，\( n_1=8 \)）  
数据：15.0, 14.5, 15.2, 15.5, 14.8, 15.1, 15.2, 14.8  
- 样本均值：\( \bar{x}_1 = \frac{15.0+14.5+15.2+15.5+14.8+15.1+15.2+14.8}{8} = 15.0125 \)  
- 离均差平方和：\( SS_1 = \sum (x_i - \bar{x}_1)^2 = 0.66875 \)  
- 样本方差：\( s_1^2 = \frac{SS_1}{n_1-1} = \frac{0.66875}{7} \approx 0.0955 \)  
#### （2）乙车床（样本2，\( n_2=9 \)）  
数据：15.2, 15.0, 14.8, 15.2, 15.0, 15.0, 14.8, 15.1, 14.8  
- 样本均值：\( \bar{x}_2 = \frac{15.2+15.0+14.8+15.2+15.0+15.0+14.8+15.1+14.8}{9} \approx 14.9889 \)  
- 离均差平方和：\( SS_2 = \sum (x_i - \bar{x}_2)^2 \approx 0.2089 \)  
- 样本方差：\( s_2^2 = \frac{SS_2}{n_2-1} = \frac{0.2089}{8} \approx 0.0261 \)  
### **3. 计算F统计量**
F检验的统计量为**较大方差除以较小方差**（确保F≥1，方便查临界值）：  
\[
F = \frac{s_1^2}{s_2^2} = \frac{0.0955}{0.0261} \approx 3.659
\]  
### **4. 确定临界值**
- 分子自由度：\( df_1 = n_1 - 1 = 7 \)  
- 分母自由度：\( df_2 = n_2 - 1 = 8 \)  
- 显著性水平 \( \alpha = 0.05 \)，双侧检验，临界值为：  
  - 上侧临界值：\( F_{\alpha/2}(df_1, df_2) = F_{0.025}(7, 8) = 4.53 \)（查F分布表）  
  - 下侧临界值：\( F_{1-\alpha/2}(df_1, df_2) = F_{0.975}(7, 8) = \frac{1}{F_{0.025}(8, 7)} = \frac{1}{4.90} \approx 0.204 \)（利用F分布性质：\( F_{1-\alpha}(df_1, df_2) = \frac{1}{F_{\alpha}(df_2, df_1)} \)）  
### **5. 决策规则**
- 若 \( F < F_{0.975}(7, 8) \) 或 \( F > F_{0.025}(7, 8) \)，拒绝 \( H_0 \)；  
- 若 \( F_{0.975}(7, 8) \leq F \leq F_{0.025}(7, 8) \)，不拒绝 \( H_0 \)。  
### **6. 结论**
计算的 \( F = 3.659 \)，介于 \( 0.204 \) 和 \( 4.53 \) 之间，**不拒绝原假设 \( H_0 \)**。因此，在 \( \alpha = 0.05 \) 的显著性水平下，认为乙车床与甲车床生产的滚珠直径方差相等。
**答案：** 乙车床生产的滚珠直径方差与甲车床相等（不拒绝 \( H_0 \)）。
